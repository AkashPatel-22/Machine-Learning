```python
# Machine Learning.
# Machine learning is part of Computer Science where we teach computers to learn from data,
# finds patterns & make decision or predictions.

# three important words:

# 1. Features - jo information model ko di jati hai.
# example - area,bedrooms, location, bathrooms, age.

# ye features hai.

# 2. Label/Target - jo cheez predict karni hai.
# e.g - price.
    # price = Target.

# 3. Model - jo features aur target ke relationship ko learn karta hai.
# featurs -> model -> prediction
```

```python
# what is AI - Artificial Intelligence
# - Computer ko aise kaam karwana jo normally human intelligence require karte hain.
#  AI - Machine ko intelligent behaviour dena

# Machine Learning = Data se patterns learn karke prediction/decision karna.
```

```python
# Machine Learning types.

Machine Learning
│
├── Supervised Learning
│
├── Unsupervised Learning
│
└── Reinforcement Learning
```

```python
Supervised Learning - Data ke saath answers available hain.
    Data ke saath answers available hain.
    Input → Answer

# Example
House features → House price
Email → Spam/Not Spam
Student data → Pass/Fail

Unsupervised Learning - Data hai, answers nahi hain.
    Input → ?
Model khud patterns/groups find karta hai.

 # Example
 Shopping customers: 
Customer A → ₹10,000/month
Customer B → ₹12,000/month
Customer C → ₹11,000/month

Customer X → ₹80,000/month
Customer Y → ₹90,000/month

Model discover kar sakta hai:
GROUP 1 → Low spending
GROUP 2 → High spending

Isko Clustering bolte hain.


Reinforcement Learning:
Model actions leta hai aur rewards/penalties se learn karta hai.

# example
🎮 Game-playing AI
    
Action
  ↓
Game
  ↓
Reward +10

wrong action:
    
Action
  ↓
Game
  ↓
Penalty -10
```
